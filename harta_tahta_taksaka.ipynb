```python
import os
import glob
import pandas as pd
from sqlalchemy import create_engine
from urllib.parse import quote_plus

# Konfigurasi Koneksi Database dan Folder Dataset
SERVER_NAME = r'ANDRIANSAH\SQLEXPRESS'
DATABASE_NAME = 'gelar_rasa_2026'
FOLDER_PATH = r'C:\Lokal LOQ\Lomba\Gelar Rasa UPNVJT 2026\Dataset DA Gelar Rasa 2026'

connection_string = (
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={SERVER_NAME};"
    f"DATABASE={DATABASE_NAME};"
    f"Trusted_Connection=yes;"
)

quoted_conn = quote_plus(connection_string)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={quoted_conn}")

# Mencari Semua File CSV di Dalam Folder
search_path = os.path.join(FOLDER_PATH, '*.csv')
all_csv_files = glob.glob(search_path)

print(f"Ditemukan {len(all_csv_files)} file CSV di dalam folder.\n")

# Looping untuk Membaca dan Memasukkan Setiap File ke SQL Server
for file_path in all_csv_files:
    file_name = os.path.basename(file_path)
    table_name = os.path.splitext(file_name)[0]
    
    # Membersihkan karakter pada nama tabel
    table_name = table_name.replace('-', '_').replace(' ', '_').replace('.', '_')
    
    try:
        # Mendeteksi pemisah secara otomatis
        df = pd.read_csv(file_path, sep=None, engine='python', on_bad_lines='skip')
        
        df.to_sql(
            name=table_name, 
            con=engine, 
            if_exists='replace', 
            index=False,
            chunksize=1000
        )
        
        print(f"Berhasil: {file_name} -> Tabel [{table_name}] ({len(df)} baris)")
        
    except Exception as e:
        print(f"Gagal mengimpor {file_name}. Error: {e}\n")

print(r"Semua proses impor selesai!")
```

Data dengan qc_flag CALIBRATING dan MISSING diubah menjadi NaN agar diabaikan dalam perhitungan *downsampling* sehingga hasilnya merepresentasikan udara ambien asli, sedangkan data SUSPECT tetap dipertahankan karena lonjakan ekstremnya berpotensi merekam kejadian polusi nyata seperti asap Karhutla yang memerlukan investigasi lebih lanjut.

```python
# Mengubah fitur Polutan dan Cuaca CALIBRATING dan MISSING menjadi NaN pada data 15 menit
kondisi_rusak_15 = plb_01_15['qc_flag'].isin(['MISSING', 'CALIBRATING'])
plb_01_15.loc[kondisi_rusak_15, features] = np.nan

# Mengubah fitur Polutan dan Cuaca MISSING menjadi NaN pada data 60 menit
kondisi_rusak_60 = plb_01_60['qc_flag'].isin(['MISSING'])
plb_01_60.loc[kondisi_rusak_60, features] = np.nan

# Memastikan perubahan berhasil
plb_01_15.groupby('qc_flag')[features].apply(lambda x: x.isnull().sum())
```

```python
# Fungsi QC menggunakan aturan 75% validitas dari observasi duration_minutes 15 menit untuk menghasilkan QC pada resolusi 60 menit
def get_advanced_hourly_qc(x):
    valid_count = (x == 'VALID').sum()
    if valid_count >= 3: 
        return 'VALID'
    elif (x == 'SUSPECT').any():
        return 'SUSPECT'
    elif (x == 'CALIBRATING').sum() > 0:
        return 'CALIBRATING'
    else:
        return 'MISSING'

# Agregasi pada duration_minutes 15 menit
agg_rules = {
    'measurement_id': 'first',
    'sensor_id': 'first',
    'timestamp_end': 'max',
    'co_ug_m3': 'mean',
    'humidity_pct': 'mean',
    'no2_ug_m3': 'mean',
    'o3_ug_m3': 'mean',
    'pm10_ug_m3': 'mean',
    'pm25_ug_m3': 'mean',
    'so2_ug_m3': 'mean',
    'temperature_c': 'mean',
    'wind_speed_ms': 'mean',
    'wind_direction_deg': lambda x: x.mode()[0] if not x.dropna().empty else np.nan,
    'qc_flag': get_advanced_hourly_qc,
    'duration_minutes': lambda x: 60
}

plb_01_15_resampled = (
    plb_01_15.set_index('timestamp_start')
    .resample('60min')
    .agg(agg_rules)
    .dropna(how='all')
    .reset_index()
)
plb_01_15_resampled = plb_01_15_resampled[
    plb_01_15_resampled['measurement_id'].notna()
].reset_index()
plb_01_15_resampled
```

# **Data Cleaning**

```python
# Polutan
polutan = ['co_ug_m3', 'no2_ug_m3', 'o3_ug_m3', 'pm10_ug_m3', 'pm25_ug_m3', 'so2_ug_m3']

# Obervasi invalid per stasiun dari merged_data
grouped = merged_data.groupby(['table_name', 'sensor_id', 'date'])[polutan]

daily_count = grouped.count().reset_index()
daily_mean = grouped.mean().reset_index()

# Mengecek baris dengan minimal satu polutan invalid dengan jumlah observasi < 18 per hari
invalid_days = daily_count[daily_count[polutan].lt(18).any(axis=1)]
invalid_days
```

```python
# Stasiun dengan observasi < 18 per hari
observations_under_18 = (
    daily_count.set_index(['table_name', 'date'])[polutan]
    .lt(18)
    .groupby('table_name')
    .sum()
)

# Stasiun dengan setidaknya satu polutan yang memiliki observasi < 18 per hari
observations_under_18_filtered = observations_under_18[
    observations_under_18[polutan].gt(0).any(axis=1)
]

observations_under_18_filtered
```

Berdasarkan hasil eksplorasi agregrasi data MDN_01_Medan_AQMS, MER_01_Merauke_AQMS, PLB_01_TalangBetutu_AQMS harian terdapat timestamp yang hilang, sehingga menyebabkan jumlah observarsi harian < 18, maka dapat disimpulkan data tersebut menjadi invalid ketika dilakukan agregasi data harian.

```python
# Membuat kerangka waktu penuh 24 jam
invalid_keys = (
    invalid_days[
        ['table_name', 'sensor_id', 'date']
    ]
    .drop_duplicates()
)

full_grid = (
    invalid_keys.assign(
        timestamp_start=lambda d: pd.to_datetime(d['date']).apply(
            lambda dt: pd.date_range(
                dt,
                periods=24,
                freq='h'
            )
        )
    )
    .explode('timestamp_start')
    .reset_index(drop=True)
)

# Menggabungkan kerangka 24 jam dengan data aktual
df_full_check = (
    full_grid.merge(
        merged_data,
        on=[
            'table_name',
            'sensor_id',
            'date',
            'timestamp_start'
        ],
        how='left'
    )
)

# Menandai setiap jam yang minimal satu polutannya hilang
df_full_check['is_missing'] = (
    df_full_check[polutan]
    .isna()
    .any(axis=1)
)

# Mengelompokkan blok kosong berurutan secara kronologis
missing_hours = (
    df_full_check[
        df_full_check['is_missing']
    ]
    .copy()
)

missing_hours['gap_block'] = (
    missing_hours
    .groupby(
        ['table_name', 'sensor_id', 'date']
    )['timestamp_start']
    .diff()
    .ne(pd.Timedelta(hours=1))
    .cumsum()
)

gap_blocks = (
    missing_hours
    .groupby(
        [
            'table_name',
            'sensor_id',
            'date',
            'gap_block'
        ]
    )
    .agg(
        empty_start=('timestamp_start', 'min'),
        empty_end=(
            'timestamp_start',
            lambda x: x.max() + pd.Timedelta(hours=1)
        ),
        gap_duration_hours=('timestamp_start', 'count'),
        polutan=(
            'is_missing',
            lambda x: [
                col for col in polutan
                if missing_hours.loc[x.index, col]
                .isna()
                .all()
            ]
        )
    )
    .reset_index()
    .drop(columns='gap_block')
)

gap_blocks
```

```python
# gap_duration_hours
for table_name, data in gap_blocks.groupby('table_name', sort=True):
    gap_duration_hours = (
        data.groupby('gap_duration_hours')
        .size()
        .reset_index(name='total_data')
        .sort_values('gap_duration_hours', ascending=False)
        .reset_index(drop=True)
    )

    print(f'\n{table_name}')
    display(gap_duration_hours)
```

```python
# Membuat grid jam untuk setiap gap pada gap_blocks
grid_records = []

for row in gap_blocks.itertuples(index=False):
    timestamps = pd.date_range(
        start=row.empty_start,
        end=row.empty_end - pd.Timedelta(hours=1),
        freq='h'
    )

    grid_records.extend({
        'table_name': row.table_name,
        'sensor_id': row.sensor_id,
        'timestamp_start': timestamp
    } for timestamp in timestamps)

grid_df = pd.DataFrame(grid_records).drop_duplicates()


# Timestamp yang belum ada di merged_data
existing_keys = (
    merged_data[
        ['table_name', 'sensor_id', 'timestamp_start']
    ]
    .copy()
)

existing_keys['timestamp_start'] = pd.to_datetime(
    existing_keys['timestamp_start']
)

grid_df['timestamp_start'] = pd.to_datetime(
    grid_df['timestamp_start']
)

new_grid = (
    grid_df
    .merge(
        existing_keys.drop_duplicates(),
        on=['table_name', 'sensor_id', 'timestamp_start'],
        how='left',
        indicator=True
    )
    .query("_merge == 'left_only'")
    .drop(columns='_merge')
)


# Membuat baris baru dengan struktur yang sama seperti merged_data
new_rows = pd.DataFrame(
    index=range(len(new_grid)),
    columns=merged_data.columns
)

new_rows['table_name'] = new_grid['table_name'].to_numpy()
new_rows['sensor_id'] = new_grid['sensor_id'].to_numpy()
new_rows['timestamp_start'] = new_grid['timestamp_start'].to_numpy()

if 'timestamp_end' in new_rows.columns:
    new_rows['timestamp_end'] = (
        new_rows['timestamp_start'] + pd.Timedelta(hours=1)
    )

if 'date' in new_rows.columns:
    new_rows['date'] = (
        new_rows['timestamp_start']
        .dt.normalize()
    )

if 'qc_flag' in new_rows.columns:
    new_rows['qc_flag'] = 'MISSING'

if 'duration_minutes' in new_rows.columns:
    new_rows['duration_minutes'] = 60


# Menambahkan baris grid ke data
merged_data = (
    pd.concat([merged_data, new_rows], ignore_index=True)
    .sort_values(
        ['table_name', 'sensor_id', 'timestamp_start']
    )
    .reset_index(drop=True)
)
merged_data
```

```python
# Missing value setiap table_name group by qc_flag dan fitur Polutan dan Cuaca
for table_name in merged_data['table_name'].unique():

    missing_value = (
        merged_data.loc[merged_data['table_name'] == table_name]
        .groupby('qc_flag')[features]
        .apply(lambda x: x.isnull().sum())
    )

    missing_value = missing_value[
        (missing_value > 0).any(axis=1)
    ]

    if len(missing_value) > 0:
        print(f"\n{table_name}")
        display(missing_value)
```

```python
# Missing value overall
missing_value_overall = (
    merged_data
    .groupby('table_name')
    .apply(
        lambda df: pd.Series({
            'Total Rows': len(df),
            'Rows with Missing Value': df[features].isnull().any(axis=1).sum(),
            'Rows with Missing Value (%)': round(
                df[features].isnull().any(axis=1).mean() * 100,
                2
            ),
            'Rows Complete': (
                len(df)
                - df[features].isnull().any(axis=1).sum()
            ),
            'Rows Complete (%)': round(
                (
                    1
                    - df[features].isnull().any(axis=1).mean()
                ) * 100,
                2
            )
        })
    )
    .reset_index()
)

# table_name yang memiliki missing value
missing_value_overall = missing_value_overall[
    missing_value_overall['Rows with Missing Value'] > 0
]

# Mengubah kolom count menjadi integer
int_cols = [
    'Total Rows',
    'Rows with Missing Value',
    'Rows Complete'
]

missing_value_overall[int_cols] = (
    missing_value_overall[int_cols]
    .astype(int)
)

# Mengurutkan tabel dari yang paling banyak missing value
missing_value_overall = (
    missing_value_overall
    .sort_values(
        'Rows with Missing Value (%)',
        ascending=False
    )
    .reset_index(drop=True)
)

missing_value_overall
```

```python
# Loop table_name dari missing_value_overall
for table_name in missing_value_overall['table_name'].unique():
    table_df = merged_data[merged_data['table_name'] == table_name].sort_values('timestamp_start').copy()
    if table_df.empty:
        continue
        
    records = []
    
    # Fitur Polutan & Cuaca
    for feat in features:
        if feat not in table_df.columns:
            continue
            
        is_null = table_df[feat].isnull()
        if not is_null.any():
            continue
            
        # Mengelompokkan blok NaN yang beruntun
        null_groups = (is_null != is_null.shift()).cumsum()
        null_blocks = table_df[is_null].groupby(null_groups[is_null])
        
        for _, group in null_blocks:
            duration_hours = len(group)
            
            records.append({
                'gap_duration_hours': duration_hours,
                'feature': feat
            })
            
    if records:
        rec_df = pd.DataFrame(records)
        
        # Pivot Table
        matrix_df = (
            rec_df
            .pivot_table(
                index='gap_duration_hours',
                columns='feature',
                aggfunc='size',
                fill_value=0
            )
            .sort_index(ascending=False)
        )
        
        print(f'\n{table_name}')
        display(matrix_df)
```

```python
# Loop table_name langsung dari missing_value_overall
for table_name in missing_value_overall['table_name'].unique():
    table_df = merged_data[merged_data['table_name'] == table_name].sort_values('timestamp_start').copy()
    if table_df.empty:
        continue
        
    # Missing value per qc_flag & gap_duration_hours
    is_null_row = table_df[features].isnull().any(axis=1)
    if not is_null_row.any():
        continue
        
    # Mengelompokkan blok baris NaN yang beruntun
    null_groups = (is_null_row != is_null_row.shift()).cumsum()
    null_blocks = table_df[is_null_row].groupby(null_groups[is_null_row])
    
    records = []
    for _, group in null_blocks:
        duration_hours = len(group)
        qc_val = group['qc_flag'].iloc[0]
        
        records.append({
            'qc_flag': qc_val,
            'gap_duration_hours': duration_hours
        })
        
    if records:
        rec_df = pd.DataFrame(records)
        
        # Agregasi incidence dan total_data per qc_flag dan gap_duration_hours
        summary_df = (
            rec_df
            .groupby(['qc_flag', 'gap_duration_hours'], as_index=False)
            .size()
            .rename(columns={'size': 'incidence'})
        )
        
        summary_df['total_data'] = summary_df['gap_duration_hours'] * summary_df['incidence']
        
        # Mengurutkan berdasarkan qc_flag dan durasi gap terbesar
        summary_df = summary_df.sort_values(['qc_flag', 'gap_duration_hours'], ascending=[True, False])
        
        print(f'\n{table_name}')
        display(summary_df[['qc_flag', 'gap_duration_hours', 'incidence', 'total_data']])
```

Berdasarkan hasil eksplorasi data pada tabel yang mengandung *missing value* diperoleh kesimpulan penanganan *missing value* dengan imputasi data menggunakan interpolasi dilakukan pada table_name PLB_01_TalangBetutu_AQMS, MDN_01_Medan_AQMS, dan SM2_01_Samarinda_AQMS karena memiliki gap_duration yang wajar dan adanya qc_flag CALIBRATING MAINTENANCE yang menunjukkan sensor sedang dalam perbaikan kala itu, sedangkan MER_01_Merauke_AQMS akan didrop karena memiliki banyak *missing value* dan gap_duration yang tidak wajar.

## **ISPU**

```python
# Polutan yang akan digunakan untuk agregasi data harian
polutan = ['co_ug_m3', 'no2_ug_m3', 'o3_ug_m3', 'pm10_ug_m3', 'pm25_ug_m3', 'so2_ug_m3']

# Fungsi kustom untuk rata-rata dengan syarat minimal 18 observasi per hari (75% dari 24 jam)
def daily_mean_with_validity(series):
    if series.count() >= 18:
        return series.mean()
    else:
        return pd.NA

# Agregasi per stasiun, sensor, dan tanggal
merged_data_daily = (
    merged_data_gold
    .groupby(['table_name', 'sensor_id', 'date'])[polutan]
    .agg(daily_mean_with_validity)
    .reset_index()
)

merged_data_daily
```

```python
# Tabel referensi batas ISPU (Permen LHK No. 14 Tahun 2020)
# Format: (Xb, Xa, Ib, Ia)
ispu_thresholds = {
    'pm10_ug_m3': [(0, 50, 0, 50), (50, 150, 51, 100), (150, 350, 101, 200), (350, 420, 201, 300), (420, 500, 301, 500)],
    'pm25_ug_m3': [(0, 15.5, 0, 50), (15.5, 55.4, 51, 100), (55.4, 150.4, 101, 200), (150.4, 250.4, 201, 300), (250.4, 500, 301, 500)],
    'so2_ug_m3': [(0, 52, 0, 50), (52, 180, 51, 100), (180, 400, 101, 200), (400, 800, 201, 300), (800, 1200, 301, 500)],
    'co_ug_m3': [(0, 4000, 0, 50), (4000, 8000, 51, 100), (8000, 15000, 101, 200), (15000, 30000, 201, 300), (30000, 45000, 301, 500)],
    'o3_ug_m3': [(0, 120, 0, 50), (120, 235, 51, 100), (235, 400, 101, 200), (400, 800, 201, 300), (800, 1000, 301, 500)],
    'no2_ug_m3': [(0, 80, 0, 50), (80, 200, 51, 100), (200, 1130, 101, 200), (1130, 2260, 201, 300), (2260, 3000, 301, 500)]
}

# Fungsi interpolasi ISPU
def hitung_ispu(val, polutan):

    if pd.isna(val) or val < 0:
        return np.nan

    thresholds = ispu_thresholds.get(polutan)

    if thresholds is None:
        return np.nan

    for Xb, Xa, Ib, Ia in thresholds:

        if Xb <= val <= Xa:

            if Xa == Xb:
                return np.nan

            ispu = ((Ia - Ib) / (Xa - Xb)) * (val - Xb) + Ib
            return round(ispu)

    # Nilai melebihi batas maksimum tabel
    Xb, Xa, Ib, Ia = thresholds[-1]

    if Xa == Xb:
        return np.nan

    ispu = ((Ia - Ib) / (Xa - Xb)) * (val - Xb) + Ib

    return round(ispu)


# Menghitung ISPU untuk setiap polutan
for p in polutan:

    col_ispu = f"ispu_{p.replace('_ug_m3', '')}"

    merged_data_daily[col_ispu] = (
        merged_data_daily[p]
        .apply(lambda x: hitung_ispu(x, p))
        .astype(float)
    )


# Kolom ISPU
ispu_columns = [
    f"ispu_{p.replace('_ug_m3', '')}"
    for p in polutan
]


# Menampilkan hasil
display_cols = ['table_name']

if 'sensor_id' in merged_data_daily.columns:
    display_cols.append('sensor_id')

display_cols.extend(['date'])
display_cols.extend(ispu_columns)

display(
    merged_data_daily[display_cols]
)
```

```python
# Nilai ISPU tertinggi (ISPU Final) per hari
merged_data_daily['ispu_max'] = (
    merged_data_daily[ispu_columns]
    .max(axis=1)
)

# Parameter pencemar kritis
merged_data_daily['parameter_kritis'] = (
    merged_data_daily[ispu_columns]
    .idxmax(axis=1)
    .str.replace('ispu_', '', regex=False)
    .str.upper()
)

# Baris yang seluruh ISPU-nya NaN -> parameter_kritis juga NaN
merged_data_daily.loc[
    merged_data_daily[ispu_columns].isna().all(axis=1),
    'parameter_kritis'
] = np.nan

# Batas rentang dan label sesuai Permen LHK No. 14 Tahun 2020
bins = [-1, 50, 100, 200, 300, np.inf]

kategori_labels = [
    'Baik',
    'Sedang',
    'Tidak Sehat',
    'Sangat Tidak Sehat',
    'Berbahaya'
]

warna_labels = [
    'Hijau',
    'Biru',
    'Kuning',
    'Merah',
    'Hitam'
]

# Memastikan ispu_max bertipe numerik
merged_data_daily['ispu_max'] = pd.to_numeric(
    merged_data_daily['ispu_max'],
    errors='coerce'
)

# Kategori ISPU
merged_data_daily['kategori_ispu'] = (
    pd.cut(
        merged_data_daily['ispu_max'],
        bins=bins,
        labels=kategori_labels
    )
    .astype(object)
)

# Status warna
merged_data_daily['status_warna'] = (
    pd.cut(
        merged_data_daily['ispu_max'],
        bins=bins,
        labels=warna_labels
    )
    .astype(object)
)

# Kolom hasil
kolom_tampil = ['table_name']

if 'sensor_id' in merged_data_daily.columns:
    kolom_tampil.append('sensor_id')

kolom_tampil.extend([
    'date',
    'ispu_max',
    'parameter_kritis',
    'kategori_ispu',
    'status_warna'
])

merged_data_daily[kolom_tampil]
```

```python
# Urutan kategori ISPU sesuai Permen LHK No. 14/2020
urutan_kategori = ['Baik', 'Sedang', 'Tidak Sehat', 'Sangat Tidak Sehat', 'Berbahaya']

# Membuat matriks frekuensi Parameter Kritis vs Kategori ISPU
ispu = pd.crosstab(
    index=merged_data_daily['parameter_kritis'].fillna('Data Invalid (<75%)'),
    columns=merged_data_daily['kategori_ispu'].fillna('Data Invalid (<75%)'),
    margins=True,
    margins_name='Total'
)

# Mengurutkan kolom sesuai hierarki kategori ISPU (jika kolomnya tersedia di data)
kolom_urut = [k for k in urutan_kategori if k in ispu.columns]
if 'Data Invalid (<75%)' in ispu.columns:
    kolom_urut.append('Data Invalid (<75%)')
kolom_urut.append('Total')

ispu = ispu[kolom_urut]
ispu
```

## **IKU**

```python
# Tahun pengamatan
merged_data_daily['year'] = (
    pd.to_datetime(merged_data_daily['date'])
    .dt.year
)

# Ringkasan IKU tahunan
iku_summary = (
    merged_data_daily
    .groupby(
        [
            'table_name',
            'station_code',
            'sensor_id',
            'year'
        ],
        as_index=False
    )
    .agg(
        mean_no2=('no2_ug_m3', 'mean'),
        mean_so2=('so2_ug_m3', 'mean'),
        ispu_max_mean=('ispu_max', 'mean'),
        hari_tidak_sehat=('ispu_max',
                          lambda x: (x > 100).sum())
    )
)

# Indeks Model Eropa
iku_summary['I_EU'] = (
    0.5 * (
        (iku_summary['mean_no2'] / 50.0)
        + (iku_summary['mean_so2'] / 45.0)
    )
)

# Konversi ke IKU
iku_summary['IKU'] = (
    100
    - ((50.0 / 0.9) * (iku_summary['I_EU'] - 0.1))
).clip(0, 100).round(2)

# Predikat IKU
bins_iku = [-1, 30, 50, 70, 90, 100]

predikat_labels = [
    'Sangat Kurang',
    'Kurang',
    'Sedang',
    'Baik',
    'Sangat Baik'
]

iku_summary['predikat_iku'] = (
    pd.cut(
        iku_summary['IKU'],
        bins=bins_iku,
        labels=predikat_labels
    )
    .astype(object)
)

# Gabung kembali
merged_data_daily = merged_data_daily.merge(
    iku_summary[
        [
            'table_name',
            'station_code',
            'sensor_id',
            'year',
            'IKU',
            'predikat_iku'
        ]
    ],
    on=[
        'table_name',
        'station_code',
        'sensor_id',
        'year'
    ],
    how='left'
)

# Hasil
display(
    iku_summary.sort_values(
        ['year', 'IKU', 'table_name'],
        ascending=[True, True, True]
    )
)
```

```python
# Agregasi rata-rata harian dengan syarat minimal 18 jam valid
pollutants = ['pm25_ug_m3', 'pm10_ug_m3', 'so2_ug_m3', 'co_ug_m3', 'o3_ug_m3', 'no2_ug_m3']
meteo = ['temperature_c', 'humidity_pct', 'wind_speed_ms']

grouped = aqms_hourly.groupby(['station_code', 'date'])
daily_mean = grouped[pollutants + meteo].mean()
daily_count = grouped[pollutants].count()

fact_aqms_daily = daily_mean.where(daily_count >= 18).reset_index()

# Pelabelan kepatuhan baku mutu harian PP No 22 Tahun 2021
fact_aqms_daily['is_exceed_pm25_pp22'] = np.where(
    fact_aqms_daily['pm25_ug_m3'].isna(), np.nan,
    (fact_aqms_daily['pm25_ug_m3'] > 55).astype(int)
)
fact_aqms_daily['status_baku_mutu_pm25'] = np.select(
    [fact_aqms_daily['pm25_ug_m3'].isna(), fact_aqms_daily['pm25_ug_m3'] > 55],
    ['Data Tidak Valid', 'Melebihi Baku Mutu'],
    default='Memenuhi Baku Mutu'
)

# Pelabelan episode kabut asap dan anomali ekstrem
p95_station = fact_aqms_daily.groupby('station_code')['pm25_ug_m3'].transform(lambda s: s.quantile(0.95))
fact_aqms_daily['episode_polusi'] = np.select(
    [
        fact_aqms_daily['pm25_ug_m3'].isna(),
        (fact_aqms_daily['pm25_ug_m3'] > p95_station) & (fact_aqms_daily['pm25_ug_m3'] > 100),
        fact_aqms_daily['pm25_ug_m3'] > 55
    ],
    ['Tidak Valid', 'Episode Ekstrem / Kabut Asap', 'Elevasi Sedang-Tinggi'],
    default='Normal / Latar'
)

# Menghitung rasio partikulat dan kondisi dispersi angin
fact_aqms_daily['pm25_pm10_ratio'] = (fact_aqms_daily['pm25_ug_m3'] / fact_aqms_daily['pm10_ug_m3']).round(3)
fact_aqms_daily['kondisi_angin'] = np.where(
    fact_aqms_daily['wind_speed_ms'] < 1.5, 'Stagnan (<1.5 m/s)', 'Dispersi Aktif (>=1.5 m/s)'
)

# Menambahkan kode rumah sakit dan fitur lag PM2.5 hingga 5 hari
fact_aqms_daily = (
    fact_aqms_daily
    .merge(station_hospital_map, on='station_code', how='left')
    .sort_values(['station_code', 'date'])
    .reset_index(drop=True)
)

for lag in range(1, 6):
    fact_aqms_daily[f'pm25_lag{lag}'] = fact_aqms_daily.groupby('station_code')['pm25_ug_m3'].shift(lag)

fact_aqms_daily
```

## **Features Importance**

```python
from sklearn.ensemble import RandomForestRegressor

# Menggabungkan data per jam dengan fitur spasial lokasi
aqms_meta = aqms_hourly.merge(
    dim_location[['station_code', 'distance_to_road_m', 'traffic_intensity_per_h', 'baseline_pm25']],
    on='station_code', how='left'
)

# Memastikan kolom numerik bertipe float64 sebelum kalkulasi trigonometri dan Random Forest
num_cols_rf = [
    'pm25_ug_m3', 'temperature_c', 'humidity_pct', 'wind_speed_ms',
    'wind_direction_deg', 'distance_to_road_m', 'traffic_intensity_per_h', 'baseline_pm25'
]
for col in num_cols_rf:
    aqms_meta[col] = pd.to_numeric(aqms_meta[col], errors='coerce')

aqms_meta['wind_dir_sin'] = np.sin(np.deg2rad(aqms_meta['wind_direction_deg']))
aqms_meta['wind_dir_cos'] = np.cos(np.deg2rad(aqms_meta['wind_direction_deg']))
aqms_meta['is_weekend'] = (aqms_meta['timestamp_start'].dt.dayofweek >= 5).astype(int)

driver_features = [
    'temperature_c', 'humidity_pct', 'wind_speed_ms', 'wind_dir_sin', 'wind_dir_cos',
    'distance_to_road_m', 'traffic_intensity_per_h', 'baseline_pm25', 'hour', 'is_weekend'
]

label_map = {
    'traffic_intensity_per_h': 'Intensitas Lalu Lintas (kendaraan/jam)',
    'distance_to_road_m': 'Jarak ke Jalan Raya (meter)',
    'baseline_pm25': 'Karakteristik Baseline Wilayah',
    'wind_speed_ms': 'Kecepatan Angin (m/s)',
    'humidity_pct': 'Kelembapan Relatif (%)',
    'temperature_c': 'Suhu Udara (°C)',
    'hour': 'Jam Pemantauan (Siklus Diurnal)',
    'wind_dir_sin': 'Arah Angin (Timur-Barat)',
    'wind_dir_cos': 'Arah Angin (Utara-Selatan)',
    'is_weekend': 'Hari Kerja vs Akhir Pekan'
}

# Melatih model Random Forest untuk melihat kontribusi setiap faktor
df_rf = aqms_meta.dropna(subset=['pm25_ug_m3'] + driver_features)
rf = RandomForestRegressor(n_estimators=100, max_depth=12, random_state=42, n_jobs=-1)
rf.fit(df_rf[driver_features], df_rf['pm25_ug_m3'])

model_pollution_drivers = pd.DataFrame({
    'fitur_kode': driver_features,
    'faktor_penentu': [label_map[c] for c in driver_features],
    'kategori_faktor': [
        'Meteorologi', 'Meteorologi', 'Meteorologi', 'Meteorologi', 'Meteorologi',
        'Aktivitas & Spasial', 'Aktivitas & Spasial', 'Aktivitas & Spasial', 'Temporal', 'Temporal'
    ],
    'importance_pct': (rf.feature_importances_ * 100).round(2)
}).sort_values('importance_pct', ascending=False).reset_index(drop=True)

model_pollution_drivers
```

## **Relative Risk**

```python
import statsmodels.api as sm
import statsmodels.formula.api as smf

# Menggabungkan data kualitas udara harian dengan kasus kesehatan rumah sakit
panel_health = fact_aqms_daily.merge(health_outcomes_daily, on=['hospital_id', 'date'], how='inner')
lag_cols = ['pm25_ug_m3'] + [f'pm25_lag{l}' for l in range(1, 6)]
lag_formula = " + ".join(lag_cols)

outcomes_map = {
    'resp_admissions_under5': ('ISPA Balita (<5 Tahun)', 'Kelompok Rentan'),
    'resp_admissions_over60': ('ISPA Lansia (>60 Tahun)', 'Kelompok Rentan'),
    'asthma_emergency_cases': ('Gawat Darurat Asma', 'Penyakit Pernapasan'),
    'copd_admissions': ('Rawat Inap PPOK', 'Penyakit Pernapasan'),
    'cvd': ('Penyakit Kardiovaskular (CVD)', 'Kardiovaskular'),
    'death': ('Kematian Harian', 'Mortalitas')
}

# Memastikan seluruh variabel numerik bertipe float64 agar tidak dianggap kategori oleh patsy
num_health_cols = list(outcomes_map.keys()) + lag_cols + ['temp_c', 'rhum_pct', 'time']
for c in num_health_cols:
    panel_health[c] = pd.to_numeric(panel_health[c], errors='coerce')

# Menghitung Relative Risk per kenaikan 10 ug/m3 PM2.5 pada Lag 0 hingga Lag 5
lag_results = []
for col, (label_penyakit, kelompok) in outcomes_map.items():
    sub = panel_health.dropna(subset=[col] + lag_cols + ['temp_c', 'rhum_pct', 'time']).copy()
    formula = f"{col} ~ {lag_formula} + temp_c + rhum_pct + C(dow) + C(station_code) + bs(time, df=6)"
    res = smf.glm(formula=formula, data=sub, family=sm.families.NegativeBinomial()).fit()
    
    conf = res.conf_int()
    for idx, lag_var in enumerate(lag_cols):
        beta = res.params[lag_var]
        ci_low, ci_high = conf.loc[lag_var]
        lag_results.append({
            'outcome_kode': col,
            'indikator_kesehatan': label_penyakit,
            'kelompok_kasus': kelompok,
            'lag_hari': idx,
            'lag_label': f"Lag {idx} Hari" if idx > 0 else "Hari H (Lag 0)",
            'rr_per_10ug': round(np.exp(beta * 10), 4),
            'rr_ci_low': round(np.exp(ci_low * 10), 4),
            'rr_ci_high': round(np.exp(ci_high * 10), 4),
            'pct_kenaikan_risiko': round((np.exp(beta * 10) - 1) * 100, 2),
            'p_value': round(res.pvalues[lag_var], 4),
            'signifikan_95': 'Signifikan' if res.pvalues[lag_var] < 0.05 else 'Tidak Signifikan'
        })

model_health_lag_rr = pd.DataFrame(lag_results)
model_health_lag_rr
```

```python
# Menghitung beban polusi dan beban kasus kelompok rentan per wilayah
panel_health['total_kasus_rentan'] = panel_health['resp_admissions_under5'] + panel_health['resp_admissions_over60']

station_summary = (
    panel_health.groupby('station_code')
    .agg(
        rata2_pm25=('pm25_ug_m3', 'mean'),
        hari_melebihi_bm=('is_exceed_pm25_pp22', 'sum'),
        total_kasus_resp=('resp', 'sum'),
        total_kasus_rentan=('total_kasus_rentan', 'sum')
    )
    .reset_index()
)

network_audit = dim_location.merge(station_summary, on='station_code', how='left')

# Menghitung skor komposit prioritas intervensi jaringan
def min_max_100(s):
    return ((s - s.min()) / (s.max() - s.min() + 1e-9)) * 100

network_audit['skor_beban_polusi'] = min_max_100(network_audit['rata2_pm25'].fillna(0))
network_audit['skor_beban_kesehatan'] = min_max_100(network_audit['total_kasus_rentan'].fillna(0))
network_audit['skor_defisit_data'] = 100 - network_audit['data_completeness'].fillna(0)

network_audit['indeks_prioritas_jaringan'] = (
    0.40 * network_audit['skor_beban_polusi'] +
    0.40 * network_audit['skor_beban_kesehatan'] +
    0.20 * network_audit['skor_defisit_data']
).round(1)

network_audit['rekomendasi_aksi'] = np.select(
    [
        (network_audit['data_completeness'] < 75),
        (network_audit['indeks_prioritas_jaringan'] >= 65) & (network_audit['data_completeness'] >= 75),
        (network_audit['indeks_prioritas_jaringan'] >= 45)
    ],
    [
        'Prioritas 1: Peremajaan Sensor & Stabilisasi Telemetri',
        'Prioritas 1: Penambahan Titik Pemantau Sekunder',
        'Prioritas 2: Optimalisasi Pemantauan Musiman'
    ],
    default='Pertahankan Pemantauan Rutin'
)

network_audit
```
